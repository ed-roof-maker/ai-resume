## Create Expert Knowledge Agent

```python
from schizophrenia import retrieve_knowledge
from schizophrenia_agents import rerank_chunks
```

```python
prefix = "Show the; Evidence, Sources, Final Assessment, Verdict, and Reasoning, for the transcript. Claim: "
question = "Does stress indirectly cause schizophrenia?"
candidates = retrieve_knowledge(
    f"{prefix}{question}",
    n_results=10
)
ranked_results = rerank_chunks(chunks=candidates)
```

```python
ranked_results
```

Invoke agent

```python
from schizophrenia_agents import llm, agent

question = "Can i just be on low dose antipsychotics?"

config = {
    "configurable": {
        "thread_id": "debate-session-1"
    }
}

response = agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": question,
            }
        ]
    },
    config=config,
)

```
