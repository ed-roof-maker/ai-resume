## Create Knowledge Research Docs Via The Debater Agent

The docs will be loaded into a vector DB, and be used by the RAG agent.

```python
from debater import debate, summarise_debate, judge_summary, transcript_to_markdown
```

```python
from pathlib import Path
import re
from time import time

def slugify(text: str, max_length: int = 80) -> str:
    """Convert text into a safe filename."""
    slug = text.lower().strip()
    slug = re.sub(r"[^\w\s-]", "", slug)
    slug = re.sub(r"[-\s]+", "-", slug)
    return slug[:max_length].strip("-")

def debate_claims_and_save_to_file(claims):
    output_dir = Path(f"ai-schizophrenia-facts-{WEAK_MODEL}-{int(time())}")
    output_dir.mkdir(parents=True, exist_ok=True)

    for index, claim in enumerate(claims, start=1):
        try:

            print(f"\n{'=' * 80}")
            print(f"Debate {index}: {claim}")
            print(f"{'=' * 80}")

            # --------------------------------------------------
            # Run debate
            # --------------------------------------------------

            transcript = debate(
                claim=claim,
                rounds=6,
            )

            # --------------------------------------------------
            # Summarise debate
            # --------------------------------------------------

            summary = summarise_debate(
                claim=claim,
                transcript=transcript,
            )

            # --------------------------------------------------
            # Judge debate
            # --------------------------------------------------
            judgement = judge_summary(
                claim=claim,
                summary=summary,
            )

            # --------------------------------------------------
            # Write debate to Markdown
            # --------------------------------------------------

            filename = f"{index:03d}-{slugify(claim)}.md"
            output_path = output_dir / filename

            markdown = f"""# Claim

{claim}


# Debate

{transcript_to_markdown(transcript)}


# Summary

{summary}


# Judgement

{judgement}
"""

            output_path.write_text(
                markdown,
                encoding="utf-8",
            )

            print(f"Saved: {output_path}")

        except Exception as exc:
            print(f"ERROR processing {claim}: {exc}")
```
