## Create Resume Knowledge Agent

```python
from digitaltwin_db import retrieve_knowledge
from digitaltwin_agents import rerank_chunks
```

```python
prefix = "Show the answer. Question: "
question = "Why is there a large gap in your resume?"
query = f"{prefix}{question}"
about = retrieve_knowledge(
    query,
    n_results=10
)
ranked_results = rerank_chunks(query, chunks=about)
```

```python
ranked_results
```

Invoke agent

```python
questions = [
    'Why are there large gaps in your resume?',
    'Do you have a family?',
    'Tell me about your medical condition.',
    'Tell me more about schizophrenia and how it affects your work.',
    'Where do you live?',
    'How can i contact you?',
    """Tell me about a time you solved a particularly difficult technical problem. 
    What was the problem, how did you investigate it, and what was the outcome?""",
]
```

```python
from digitaltwin_agents import llm, agent
from IPython.display import Markdown, display, update_display

def print_questions():
    for question in questions:
        config = {
            "configurable": {
                "thread_id": "debate-session-1"
            }
        }

        response = agent.invoke(
            {
                "messages": [
                    {
                        "role": "user",
                        "content": question,
                    }
                ]
            },
            config=config,
        )
        display(Markdown(''))
        display(Markdown(f'QUESTION: {question}'))                     
        display(Markdown(f'ANSWER: {response["messages"][-1].content}'))          

```
